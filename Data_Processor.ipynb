```python
# Standard libraries for data analysis and display
import numpy as np
import pandas as pd
from IPython.display import display
```

```python
# Read and combine data for tournament seeds, regular season results, tournament results, and 538 ratings
# Add Division column to indicate the division (M for men's, W for women's)

df_seeds = pd.concat(
    [
        pd.read_csv("raw/MNCAATourneySeeds.csv").assign(Divison="M"),
        pd.read_csv("raw/WNCAATourneySeeds.csv").assign(Divison="W"),
    ]
).reset_index(drop=True)

df_compact_season_results = pd.concat(
    [
        pd.read_csv("raw/MRegularSeasonCompactResults.csv").assign(Divison="M"),
        pd.read_csv("raw/WRegularSeasonCompactResults.csv").assign(Divison="W"),
    ]
).reset_index(drop=True)

df_compact_tourney_results = pd.concat(
    [
        pd.read_csv("raw/MNCAATourneyCompactResults.csv").assign(Divison="M"),
        pd.read_csv("raw/WNCAATourneyCompactResults.csv").assign(Divison="W"),
    ]
).reset_index(drop=True)

fivethiryeight_scores = pd.concat(
    [
        pd.read_csv("raw/538ratingsMen.csv").assign(Divison="M"),
        pd.read_csv("raw/538ratingsWomen.csv").assign(Divison="W"),
    ]
).reset_index(drop=True)


# Display the concatenated DataFrames
display(df_seeds)
print("Season Results")
display(df_compact_season_results)
print("Tournament Results")
display(df_compact_tourney_results)
print("538 Scores")
display(fivethiryeight_scores)
```

```python
# Check for missing values in each DataFrame and convert the result into a DataFrame with column names for clarity
missing_seeds = df_seeds.isnull().sum().reset_index()
missing_seeds.columns = ["Column", "Missing Values"]

missing_season_results = df_compact_season_results.isnull().sum().reset_index()
missing_season_results.columns = ["Column", "Missing Values"]

missing_tourney_results = df_compact_tourney_results.isnull().sum().reset_index()
missing_tourney_results.columns = ["Column", "Missing Values"]

missing_fivethiryeight_scores = fivethiryeight_scores.isnull().sum().reset_index()
missing_fivethiryeight_scores.columns = ["Column", "Missing Values"]

# Display the missing values for each dataset to check for completeness
print("Seeds")
display(missing_seeds)
print("\n")
print("Season Results")
display(missing_season_results)
print("\n")
print("Tournament Results")
display(missing_tourney_results)
print("\n")
print("538 Scores")
display(missing_fivethiryeight_scores)
```

```python
# Check for duplicates in each DataFrame and count the number of duplicated rows
duplicates_seeds = pd.DataFrame({
    "DataFrame": ["df_seeds"],
    "Duplicate Count": [df_seeds.duplicated().sum()]
})

duplicates_season_results = pd.DataFrame({
    "DataFrame": ["df_compact_season_results"],
    "Duplicate Count": [df_compact_season_results.duplicated().sum()]
})

duplicates_tourney_results = pd.DataFrame({
    "DataFrame": ["df_compact_tourney_results"],
    "Duplicate Count": [df_compact_tourney_results.duplicated().sum()]
})

duplicates_fivethiryeight_scores = pd.DataFrame({
    "DataFrame": ["fivethiryeight_scores"],
    "Duplicate Count": [fivethiryeight_scores.duplicated().sum()]
})

# Display the duplicate counts for each dataset to inspect whether any duplicates exist
display(duplicates_seeds)
display(duplicates_season_results)
display(duplicates_tourney_results)
display(duplicates_fivethiryeight_scores)
```

```python
# Get the data types for each DataFrame's columns
dtypes_seeds = df_seeds.dtypes.reset_index()
dtypes_seeds.columns = ["Column", "Data Type"]

dtypes_season_results = df_compact_season_results.dtypes.reset_index()
dtypes_season_results.columns = ["Column", "Data Type"]

dtypes_tourney_results = df_compact_tourney_results.dtypes.reset_index()
dtypes_tourney_results.columns = ["Column", "Data Type"]

dtypes_fivethiryeight_scores = fivethiryeight_scores.dtypes.reset_index()
dtypes_fivethiryeight_scores.columns = ["Column", "Data Type"]

# Display the data types of each dataset
print("Seeds")
display(dtypes_seeds)
print("\n")
print("Season Results")
display(dtypes_season_results)
print("\n")
print("Tournament Results")
display(dtypes_tourney_results)
print("\n")
print("538 Scores")
display(dtypes_fivethiryeight_scores)
```

```python
# Combine the season results for both the winning and losing teams into a single DataFrame
df_team_compact_season_results = pd.concat(
    [
        df_compact_season_results[["Season", "Divison", "WTeamID", "DayNum", "WScore", "LScore"]]
        .assign(Result="W")
        .rename(columns={"WTeamID": "TeamID", "WScore": "TeamScore", "LScore": "OpponentScore"}),

        df_compact_season_results[["Season", "Divison", "LTeamID", "DayNum", "WScore", "LScore"]]
        .assign(Result="L")
        .rename(columns={"LTeamID": "TeamID", "LScore": "TeamScore", "WScore": "OpponentScore"}),
    ]
).reset_index(drop=True)
```

```python
# Display Compact Season Results
df_team_compact_season_results
```

```python
# Calculate the score difference for each game and add a new 'ScoreDifference' column
df_team_compact_season_results["ScoreDifference"] = df_team_compact_season_results["TeamScore"] - df_team_compact_season_results["OpponentScore"]
```

```python
# Create a 'Win' column where 1 indicates a win and 0 indicates a loss
df_team_compact_season_results["Win"] = (df_team_compact_season_results["Result"] == "W").astype("int")
```

```python
# # Display a few random samples from Compact Season Results after the change
df_team_compact_season_results.sample(10, random_state=100)
```

```python
# Aggregate the team season data by Season, TeamID, and Division.
agg_team_season = (
    df_team_compact_season_results.groupby(["Season", "TeamID", "Divison"])
    .agg(
        TeamScore=("TeamScore", "sum"),
        OpponentScore=("OpponentScore", "sum"),
        ScoreDifference=("ScoreDifference", "sum"),
        AverageScoreDifference=("ScoreDifference", "mean"),
        MedianScoreDifference=("ScoreDifference", "median"),
        MinScoreDifference=("ScoreDifference", "min"),
        MaxScoreDifference=("ScoreDifference", "max"),
        Wins=("Win", "sum"),
        Losses=("Result", lambda x: (x == "L").sum()),
        WinPercentage=("Win", "mean"),
    )
    .reset_index()
)
```

```python
# Display a few random samples from the Aggregated Season DataFrame
agg_team_season.sample(10, random_state=100)
```

```python
# Adding a "HighSeed" column by extracting the numerical seed value (ignoring 'a' and 'b' suffixes) to represent team strength
df_seeds["HighSeed"] = (
    df_seeds["Seed"].str.replace("a", "").str.replace("b", "").str[1:].astype("int")
)
```

```python
# Merge the "HighSeed" information from df_seeds into the aggregated team season data
agg_team_season = agg_team_season.merge(
    df_seeds, on=["Season", "TeamID", "Divison"], how="left"
)
```

```python
# Display a few random samples from the Aggregated Season DataFrame after the change
agg_team_season.sample(10, random_state=100)
```

```python
# Combining tournament results for both winners and losers into a single dataframe
# Each result is assigned a "GameResult" (W for winner, L for loser), and columns are renamed for consistency
df_team_tourney_results = pd.concat(
    [
        df_compact_tourney_results[
            ["Season", "Divison", "WTeamID", "LTeamID", "WScore", "LScore"]
        ]
        .assign(GameResult="W")
        .rename(
            columns={
                "WTeamID": "TeamID",
                "LTeamID": "OpponentTeamID",
                "WScore": "TeamScore",
                "LScore": "OpponentScore",
            }
        ),
        df_compact_tourney_results[
            ["Season", "Divison", "LTeamID", "WTeamID", "LScore", "WScore"]
        ]
        .assign(GameResult="L")
        .rename(
            columns={
                "LTeamID": "TeamID",
                "WTeamID": "OpponentTeamID",
                "LScore": "TeamScore",
                "WScore": "OpponentScore",
            }
        ),
    ]
).reset_index(drop=True)
```

```python
# Creating a binary "Win" column where 1 represents a win and 0 represents a loss
df_team_tourney_results["Win"] = (df_team_tourney_results["GameResult"] == "W").astype(
    "int"
)
```

```python
# Display a few random samples from the Tournament Results DataFrame after the changes
df_team_tourney_results.sample(10, random_state=100)
```

```python
# Merge tournament results with aggregated season data for both teams and opponents
## to add features like win percentage, median score difference, and high seed
### for both teams and their opponents

df_historic_tourney_features = df_team_tourney_results.merge(
    agg_team_season[
        ["Season", "Divison", "TeamID", "WinPercentage", "MedianScoreDifference", "HighSeed"]
    ],
    on=["Season", "Divison", "TeamID"],
    how="left",
).merge(
    agg_team_season[
        ["Season", "Divison", "TeamID", "WinPercentage", "MedianScoreDifference", "HighSeed"]
    ].rename(
        columns={
            "TeamID": "OpponentTeamID",
            "WinPercentage": "OpponentWinPercentage",
            "MedianScoreDifference": "OpponentMedianScoreDifference",
            "HighSeed": "OpponentHighSeed",
        }
    ),
    on=["Season", "Divison", "OpponentTeamID"],
)
```

```python
# Display the Features DataFrame
df_historic_tourney_features
```

```python
# Calculate feature differences between teams and their opponents

df_historic_tourney_features["WinPercentageDifference"] = (
    df_historic_tourney_features["WinPercentage"]
    - df_historic_tourney_features["OpponentWinPercentage"]
)

df_historic_tourney_features["HighSeedDifference"] = (
    df_historic_tourney_features["HighSeed"]
    - df_historic_tourney_features["OpponentHighSeed"]
)

df_historic_tourney_features["MedianScoreDifferenceDifference"] = (
    df_historic_tourney_features["MedianScoreDifference"]
    - df_historic_tourney_features["OpponentMedianScoreDifference"]
)
```

```python
# Display the Features DataFrame after adding new features
df_historic_tourney_features.sample(10, random_state=100)
```

```python
# Display the 538 Ratings DataFrame
fivethiryeight_scores
```

```python
# Merge 538 ratings for teams and their opponents into the tournament features
# Remove the "TeamName" column and handle missing values for 538 ratings

# Drop "TeamName" column before merging
fivethiryeight_scores_cleaned = fivethiryeight_scores.drop("TeamName", axis=1)

# Merge team 538 ratings
df_historic_tourney_features = df_historic_tourney_features.merge(
    fivethiryeight_scores_cleaned,
    on=["Season", "Divison", "TeamID"],
    how="left",
).dropna(subset=["538rating"])

# Merge opponent 538 ratings
df_historic_tourney_features = df_historic_tourney_features.merge(
    fivethiryeight_scores_cleaned.rename(
        columns={"TeamID": "OpponentTeamID"}
    ),
    on=["Season", "Divison", "OpponentTeamID"],
    how="left",
    suffixes=("", "Opponent"),
).dropna(subset=["538ratingOpponent"])
```

```python
# Display the Features DataFrame after adding the 538 features
df_historic_tourney_features.sample(10, random_state=100)
```

```python
# Calculate the difference in 538 ratings between teams and their opponents
df_historic_tourney_features["538rating_Difference"] = (
    df_historic_tourney_features["538rating"]
    - df_historic_tourney_features["538ratingOpponent"]
)
```

```python
# Display a few random samples from the Features DataFrame after adding new features
df_historic_tourney_features.sample(10, random_state=100)
```

```python
# Display only 538 related features
df_historic_tourney_features[["Season", "TeamID", "538rating", "538ratingOpponent", "538rating_Difference"]].sample(10, random_state=100)
```

```python
# Display first few samples from the Features DataFrame
df_historic_tourney_features.head()
```

```python
# Create a baseline prediction based on team and opponent seed rankings

# If the team's seed is higher than the opponent's, assign True to BaselinePred
df_historic_tourney_features["BaselinePred"] = (
    df_historic_tourney_features["HighSeed"]
    < df_historic_tourney_features["OpponentHighSeed"]
)

# If seeds are equal, compare win percentages to set the BaselinePred value
df_historic_tourney_features.loc[
    df_historic_tourney_features["HighSeed"]
    == df_historic_tourney_features["OpponentHighSeed"],
    "BaselinePred",
] = (
    df_historic_tourney_features["WinPercentage"]
    > df_historic_tourney_features["OpponentWinPercentage"]
)
```

```python
# Display first few samples from the Features DataFrame after adding BaselinePred
df_historic_tourney_features.head()
```

```python
from sklearn.metrics import accuracy_score, log_loss
import pandas as pd

cv_scores_baseline = []
log_losses_baseline = []
seasons = []

# Loop through each season to calculate the metrics
for season in df_historic_tourney_features["Season"].unique():
    pred = df_historic_tourney_features.query("Season == @season")[
        "BaselinePred"
    ].astype("int")  # Baseline predictions for the current season

    y = df_historic_tourney_features.query("Season == @season")["Win"]  # Actual results

    # Calculate accuracy and log loss for the current season
    score = accuracy_score(y, pred)
    score_ll = log_loss(y, pred)

    # Store the accuracy, log loss, and season
    cv_scores_baseline.append(score)
    log_losses_baseline.append(score_ll)
    seasons.append(season)

# Create a DataFrame to display the results
results_df = pd.DataFrame({
    'Season': seasons,
    'Accuracy': cv_scores_baseline,
    'Log Loss': log_losses_baseline
})

# Display the DataFrame
display(results_df)

# Print the overall average accuracy across all seasons
print(f"\nBaseline accuracy: {np.mean(cv_scores_baseline):0.4f}")
```

```python
# Combine Teams data
df_all_teams = pd.concat(
    [
        pd.read_csv("raw/MTeams.csv").assign(Divison="M"),
        pd.read_csv("raw/WTeams.csv").assign(Divison="W"),
    ]
).reset_index(drop=True).drop_duplicates()

# Create a cross join to produce every possible pair
df_all_teams['key'] = 1
df_team_pairs = df_all_teams[['TeamID', 'Divison', 'key']].merge(
    df_all_teams[['TeamID', 'Divison', 'key']],
    on='key',
    suffixes=('', 'Opponent')
).query("TeamID != TeamIDOpponent").drop('key', axis=1)

df_team_pairs.rename(columns={"TeamIDOpponent": "OpponentTeamID"}, inplace=True)

# Remove pairs where TeamID > OpponentTeamID
df_team_pairs = df_team_pairs[df_team_pairs["TeamID"] < df_team_pairs["OpponentTeamID"]].copy()

# Add Season and Matchup
df_team_pairs['Season'] = 2025
df_team_pairs['Matchup'] = (
    df_team_pairs['Season'].astype(str) + "_" +
    df_team_pairs['TeamID'].astype(str) + "_" +
    df_team_pairs['OpponentTeamID'].astype(str)
)
```

```python
display(df_team_pairs[["TeamID", "OpponentTeamID", "Season", "Matchup"]])
```

```python
# Merge for "TeamID"
df_team_pairs = df_team_pairs.merge(
    agg_team_season[
        ["Season", "Divison", "TeamID", "WinPercentage", "MedianScoreDifference", "HighSeed"]
    ],
    on=["Season", "Divison", "TeamID"],
    how="left"
)

# Merge for "OpponentTeamID"
df_team_pairs = df_team_pairs.merge(
    agg_team_season[
        ["Season", "Divison", "TeamID", "WinPercentage", "MedianScoreDifference", "HighSeed"]
    ].rename(columns={
        "TeamID": "OpponentTeamID",
        "WinPercentage": "OpponentWinPercentage",
        "MedianScoreDifference": "OpponentMedianScoreDifference",
        "HighSeed": "OpponentHighSeed"
    }),
    on=["Season", "Divison", "OpponentTeamID"],
    how="left"
).reset_index(drop=True)

# Remove irrelevant columns
df_team_pairs = df_team_pairs[
    [
        "TeamID",
        "HighSeed",
        "OpponentTeamID",
        "OpponentHighSeed",
        "Season",
        "Matchup",
        "WinPercentage",
        "MedianScoreDifference",
        "OpponentWinPercentage",
        "OpponentMedianScoreDifference"
    ]
]
```

```python
display(df_team_pairs)
```

```python
# Aggregate 538 ratings
agg_538 = (
    fivethiryeight_scores_cleaned.groupby(["TeamID", "Divison"])
    .agg(
        Average538rating=("538rating", "mean"),
    )
    .reset_index()
)
```

```python
display(agg_538)
```

```python
# Merge 538 rating scores for TeamID
df_team_pairs = df_team_pairs.merge(
    agg_538,
    on=["TeamID"],
    how="left",
)

# Merge 538 rating scores for OpponentTeamID
df_team_pairs = df_team_pairs.merge(
    agg_538.rename(
        columns={"TeamID": "OpponentTeamID"}
    ),
    on=["OpponentTeamID"],
    how="left",
    suffixes=("", "Opponent"),
)
```

```python
display(df_team_pairs)
```

```python
# Fill High before Feature Calculations
for col in df_team_pairs.columns:
    if col in ["HighSeed", "OpponentHighSeed"]:
        df_team_pairs[col] = df_team_pairs[col].fillna(17)
    if col in ["Average538rating", "Average538ratingOpponent"]:
        df_team_pairs[col] = df_team_pairs[col].fillna(50)
```
